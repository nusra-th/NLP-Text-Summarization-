```python
from transformers import pipeline

summarizer = pipeline(
    task="summarization",
    model="facebook/bart-large-cnn",
    framework="pt",   # 🔥 FORCE PYTORCH
    device=-1         # CPU
)
```

```python
text = sample["article"]

summary = summarizer(
    text,
    max_length=130,
    min_length=40,
    do_sample=False
)

print("MODEL SUMMARY:\n", summary[0]["summary_text"])
print("\nACTUAL SUMMARY:\n", sample["highlights"])
```
